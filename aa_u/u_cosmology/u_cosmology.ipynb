## Special Relativity
The rules of Newtonian dynamics are predicated on the idea that all inertial observers agree on simultaneity. That is, if two events happen at the same time according to (inertial) Bob, then (inertial) Alice will agree that they happened simultaneously. Although it may be “obvious” that this should always be the case, it turns out to be incompatible with how light behaves. 

According to Maxwell’s theory of light, light waves can travel in vacuum. The reason for this is that a changing electric field $dE/dt$ creates a magnetic field $B$ and a changing magnetic field $dB/dt$ creates an electric field $E$. This back-and-forth $E\leftrightarrow B$ allows the wave to propagate by itself, with no need for a medium. 

Actually, the change of the electric field in time makes a change in the magnetic field in space and vice versa. In order to relate a time to a distance, you need a speed. Maxwell’s equations imply that the wave that you get propagates through empty space at a specific speed $c=3.0\times 10^8\,\mathrm{m/s}$. According to these same equations, when they do so, they must "oscillate" in time as they propagate forward. 

<img src="images/em_wave.png" width="300" align="right">

Einstein realized that this meant that he should never be able to catch up to light as it propagates because, if he could, he would perceive the electromagnetic field as not moving and not oscillating relative to him: the wave would be frozen in time and space. Since this violates Maxwell’s equations, it must be impossible to catch up to light as it propagates in empty space. 

If true, then that is a problem for our Newtonian understanding: Any acceleration will eventually bring us from rest to any speed we like–including speeds faster than light, so what’s going on? Einstein thought a lot about this and realized that if Maxwell is correct, Bob and Alice would both calculate the speed of light in vacuum to give the same answer. This has become the fundamental principle of special relativity: 

<center> <i>Inertial observers will always measure the same value for the speed of light in vacuum.</i></center>
<br>

Pragmatically, the speed of light in air is very close to that in vacuum, so we will generally make the approximation that the presence of air does not affect speed of light measurements.

The implications of result (1) are truly bizarre. It says that, according to Alice, Bob’s time progresses more slowly than it should. One second on Bob’s clock is measured to be $\gamma$ seconds on Alice’s: Bob’s clocks are running *slow*, and the faster Alice goes, the more Bob’s clocks slow down! Time dilation is not an illusion: If you put an atomic clock on a supersonic plane and fly it around and then compare to one you left on the ground, the one that was made to go fast is found to be running behind. If this is not shocking to you, consider it instead in terms of the speed of *life* slowing down: Biological entities age less fast at relativistic speeds.

### Spacetime
The formula for time dilation has a counterpart for distance contraction. If $L_0$ is the length measured in the object's rest frame and $L$ is the length measured by an observer who sees the object moving at speed $v$, then
$$L = L_0\sqrt{1-\frac{v^2}{c^2}}=\frac{L_0}{\gamma}\quad(2)$$
Thus distances along the direction of motion appear shorter than they do at rest, this time by a factor of $1/\gamma$. Both equations (1) and (2) are special cases of a collection of formulae that relate the coordinates of space and time of two inertial observers:
$$
\begin{aligned}
    t' &= \gamma\left( t-\frac{v}{c^2}x\right)\\
    x' &= \gamma\left( x-vt\right)\quad\quad\quad(3)
\end{aligned}
$$
In these expressions the primed observer (Alice) is traveling at a speed $v>0$ in the $+x$-direction relative to the unprimed observer (Bob). For simplicity, we are ignoring the other two spatial directions, but nothing interesting is going on in these directions for relative motion is in the $x$-direction (*i.e.* $y'=y$ and $z'=z$). These expressions describe how to transform from one inertial system of coordinates to another. They are called **Lorentz transformations**.

### Gravitational Redshift
When we drop a ball from the roof of the building, it gains kinetic energy on its way down to the sidewalk. If someone throws it back up, it loses kinetic energy on its way back to us. If this were true for light as well as our ball, that would mean that the light would be blue-shifted or red-shifted because the energy $E$ of a particle of light is determined by its frequency $f$ according to the Planck relation
$$
E=h f
$$
where $h=6.6\times 10^{-34}\,\mathrm{J\,s}$ is Planck’s constant. 

Einstein suspected this would happen, but he had to come up with a clever way to derive the effect. For a particle with a mass $m$, this is easy: The energy the particle gains as it falls a height $z$ is given by the change in the gravitational potential energy. Usually, we take this to be $U_g=-GM_\oplus m/R_\oplus$ where $M_\oplus$ is the mass of the Earth and $R_\oplus$ is the distance to the center, but since the height $z\ll R_\oplus$ is many orders of magnitude less than the radius of the Earth, this equation simplifies to $U_g=mgz$ where $g=GM_\oplus/R_\oplus^2=9.8\,\mathrm{m/s^2}$ is the acceleration due to gravity at the surface of the Earth.

Coming back to dropping balls, Einstein wanted to use the fact that the energy of an object changes by $mgz$ when it falls through the gravitational field, but he wanted to use it for light which has no mass. In typical Einstein fashion, he came up with the following ***[Gedankenexperiment](https://en.wikipedia.org/wiki/Thought_experiment)***: Suppose you drop a particle of mass $m$, but when it gets to the ground a distance $z$ below, it enters a special machine that converts all its mass and energy into a photon of the same energy, and sends the photon back up. When the photon gets back to the top of the roof, it is captured by another such machine that converts all its energy back into a particle of some mass $m'$. 

Einstein realized that if $m'\neq m$, it would mean that he could make a machine that generates energy for free. Perpetual motion machines violate the laws of thermodynamics, which even he was not ready to throw out. Thus, he concluded, it must be that $m'=m$. Now let’s trace through the steps and see what that means. 

* At the top of the roof, the particle has mass $m$ which is equivalent to an energy of $mc^2$.
* When the particle enters the special machine on the sidewalk, this energy has increased to $mc^2+mgz$ where $z$ is the distance from the roof to the sidewalk.
* The particle is converted to a photon with frequency $f=\frac mh(c^2 +gz)$, and sent back up to the roof.
* The photon arrives at the top of the roof with a frequency $f'$ which is possibly different from $f$. It is then converted back to a particle with mass $m'=m$ which means the frequency is $f'=\frac1hmc^2$.

### Big Bang Chronology
Running the cosmic expansion in reverse, we found that the Universe seems to have originated from a very small and dense phase in the past at a time $t \sim \frac1{H_0}$ about 14 billion years ago. In the 1960s Hawking and Penrose formulated and proved a series of *[singularity formation theorems](https://en.wikipedia.org/wiki/Penrose%E2%80%93Hawking_singularity_theorems)* that describe the conditions under which a collapse of space and time into a single point of infinitely high curvature will result. These are usually used to prove that a black hole must form, but applied to our observation of the hot dense phase in our early universe, they imply that our entire universe originated from a singularity in space and time called the **Big Bang**. 

This picture is classical, meaning that we are treating spacetime as an ordinary geometrical space of the types we have been discussing above. However, we know that at some point quantum mechanical effects must be taken into consideration. Doing so would define a theory of **Quantum Gravity**, and how to do this is an active area of research. Two main complementary approaches are:

1. Start with (classical) general relativity and try to "quantize it". This is what one does when passing from Newtonian Mechanics to Quantum Mechanics. In this approach, one tries to replace the main smooth entities predicted by general relativity (*e.g.* black hole horizons) with discrete algebraic objects (*e.g.* matrices).
2. Start with a quantum theory that is already consistent with special relativity and try to generalize it (*i.e.* add gravitation, or equivalently make it consistent with the equivalence principle). A quantum theory that is already special-relativistic is called a **Quantum Field Theory (QFT)**. All currently-known interactions <u>except</u> gravity are formulated in terms of models that are QFTs. Some of these models have special names like
    * *Quantum Electrodynamics* (QED) is the QFT describing **electrons** (and other charged particles) interacting with **photons**.
    * *Quantum Chromodynamics* (QCD) is the QFT describing the interactions between **quarks** and **gluons**.
    * *[Electroweak](https://en.wikipedia.org/wiki/Electroweak_interaction)* QFT is an extension of QED that includes the other **leptons** (muons, taus, and neutrinos) and a *Higgs boson*. In this theory, there are three force carriers in addition to the photon called the $W^+$, $W^-$, and $Z^0$ *electroweak gauge bosons*. In sharp contrast to the massless photons, these additional electroweak gauge bosons are *80 to 90 times more massive than a proton*.
    * **The Standard Model of Particle Physics** (SM for short) is a particular combination of the Electroweak theory and QCD that describes all the fundamental interactions together <u>except for gravity</u>.

<img src="images/standard_model_particles.png" width="550">

    * A **Grand Unified Theory** (GUT) is an extension of the SM that treats leptons and quarks as the same particle and unifies the gauge bosons (photons, gluons, and electroweak gauge bosons) into a single force carrier. Besides simplifying and "explaining" the structure of the Standard Model, these theories predict the existence of dark matter. However, they also generically predict the spontaneous decay of the proton (to an anti-electron and a standard-model particle called a pion), and despite many efforts to observe such [proton decay](https://en.wikipedia.org/wiki/Proton_decay), the proton has so far stubbornly refused to oblige. Since GUTs are traditional QFTs, we have a good understanding of them in general. What we do not know is *whether* the Standard Model actually is unified into a GUT and–if it is–*which* particular GUT is the correct one. 

A satisfactory formulation of Quantum Gravity promises to reveal very deep connections between particles and spacetime, but the current status of the field suggests it will be a long time before we have anything close enough to answer specific questions. Historically, however, lack of a theory has never stopped scientists from studying phenomena and building useful models, and so it will be with our work. ### Grand Unification Epoch $t_{Pl} \lesssim t \lesssim t_{GUT}$
We now exit the period of complete ignorance and enter the period of at-least-some-clue. Gravity begins to behave as a classical theory and separates from the GUT theory. Quarks and leptons interact and switch types by exchanging grand unified gauge bosons until the temperature (and therefore the energy) of the Universe drops too low for these gauge bosons to form. This epoch ends around the GUT time $t_{GUT}\sim 10^{-36}\,\mathrm{s}$.

### Electroweak Epoch $t_{GUT} \lesssim t \lesssim t_{TeV}$
Below the GUT scale, quarks and leptons begin to behave very differently. The energy drops below the mass of the GUT gauge bosons, so they cannot be created. Since these particles change quarks into leptons and *vice versa*, the symmetry between them is "broken", effectively consigning any GUT matter particle to henceforth be a specific quark or lepton. 

Whatever stable matter remains from the GUT theory that does not interact with gluons and/or photons becomes dark matter. We know of no GUT theories that do not automatically introduce such extra matter (but there are other beyond-the-standard-model paradigms that avoid them). 

### Inflation and Reheating
Around $t\sim 10^{-36}\,\mathrm{s}$ something crazy happens: The universe suddenly grows in size by a factor of $\sim e^{60}\approx 10^{26}$ in a tiny time of $t\sim 10^{-32}\,\mathrm{s}$!!! How this happens is an active field of study. There are multiple mechanisms that have been proposed to trigger inflation, but each version so far gives rise to more questions and problems for which we do not yet have answers. Nevertheless, we have ample quantitative experimental evidence that this scenario did, in fact, happen. Because we do not currently have an adequate theoretical foundation for this part of our history, it is more accurate to refer to an "inflationary paradigm" rather than a "theory". 

Next, whatever particle physics was responsible for triggering inflation is believed to have dumped the energy it had into Standard Model particles. This causes the matter and radiation in the post-inflationary Universe to *reheat*, leaving it in a state of a hot and dense quark-gluon plasma.

The inflationary paradigm implies that the universe is supercooled from $T_i\sim 10^{27}\,\mathrm{K}$ to $T_f\sim 10^{22}\,\mathrm{K}$. Regions previously in thermal equilibrium are moved out of causal contact with one another very rapidly resulting in a uniform distribution of matter and radiation throughout the entire Universe.
