## Angles

The geometry of triangles is a cornerstone of mathematics and physics. In this course, we will be using the properties of similar triangles and right triangles to determine distances and sizes. Specifically, we will be making extensive use of the trigonometric functions sine, cosine, and tangent.


## Right triangles

![Right triangle with sides A, B, C and angles alpha, beta.](images/right_triangle_abc.png)

Right triangles are triangles for which one of the three angles is $90^\circ$. Such a triangle has very nice properties. For example, it has a longest edge, which is the edge not adjacent to the right angle. In the figure above, the right angle is labeled with a corner/hook that looks like $\Gamma$. Notice that the edge labeled $C$ is longer than the other two. This edge is called the **hypotenuse** of the triangle.

Pick one of the two other angles and call it $\alpha$. In the figure above, we picked it to be the angle opposite the edge labeled $A$. Because of this, the edge $A$ is said to be **opposite** $\alpha$. Similarly, the edge labeled $B$ is said to be **adjacent**.[^adjacent-note] Given this naming, it is often useful to relabel the edges as $H$, $O$, and $A$ as in the figure below.

![Right triangle relabeled with hypotenuse H, opposite side O, adjacent side A.](images/right_triangle_hoa.png)

[^adjacent-note]: You could complain that the hypotenuse is also adjacent to that angle, but that edge is special. It is the longest. The convention is that the long edge is called the hypotenuse and the shorter one is the adjacent edge.


Now let’s look at how the lengths of the edges $H$, $O$, and $A$ depend on each other and on the angle $\alpha$. Suppose we fix the length of $A$ and make $O$ bigger or smaller. Then we can clearly see that both $\alpha$ and $H$ get bigger or smaller. We can also switch $A$ and $O$ to come to the same conclusion for $H$, but this time $\alpha$ will do the opposite. Some experimentation with graph paper and a ruler should convince you that the edges satisfy the following relationship:

$$
H = \sqrt{A^2 + O^2}.
\tag{1}
$$

This rule is called the [Theorem of Pythagoras](https://en.wikipedia.org/wiki/Pythagorean_theorem).

Similarly, the lengths of $A$ and $O$ should suffice to determine $\alpha$: just draw a triangle with some fixed sides $A$ and $O$ and then you will see that it gives some angle $\alpha$ that you can measure with a protractor. If you continue with your experiments, you will also notice that if you double both $A$ and $O$, this does not change the angle $\alpha$. This suggests that there should be a direct relationship between the angle and the ratio $O/A$. Indeed, there is a special function called the tangent function, abbreviated $\tan$, that gives this relation:

$$
\tan(\alpha) = \frac{O}{A}.
\tag{2}
$$

Similar considerations imply that there should be special functions to express the angle in terms of the lengths of the hypotenuse and one other edge. They are called the sine, abbreviated $\sin$, and cosine, abbreviated $\cos$:

$$
\sin(\alpha) = \frac{O}{H}
\qquad \text{and} \qquad
\cos(\alpha) = \frac{A}{H}.
\tag{3}
$$

## Radians

A circle of radius $r$ has a circumference $C=2\pi r$. If the radius is $r=1\,\mathrm{m}$, the circle has a circumference of $2\pi\,\mathrm{m}\approx 6.3\,\mathrm{m}$. Note that we needed to specify a unit of measure of length, but whatever units we choose, $r=1$ always gives a circumference of $2\pi$ in the same units.

If we start somewhere on a circle of radius $r$ and follow the arc of the circle to an angle of $45^\circ$, how far have we traveled? This arc length $\ell$ must be an eighth of the circumference, so $\ell=(\pi/4)r$.

Going around the circle once corresponds to going through an angle of $360^\circ$. We see that this corresponds to going through an angle of $2\pi$ in some set of units that is not degrees; we call them **radians**:

$$
360^\circ \leftrightarrow 2\pi\,\mathrm{rad},
\qquad
1^\circ \leftrightarrow \frac{\pi}{180}\,\mathrm{rad}\approx 0.0175\,\mathrm{rad},
\qquad
57.296^\circ \leftrightarrow 1\,\mathrm{rad}.
$$

Radians have the wonderful property that if an angle $\alpha$ is given in radians, then the length of an arc with this angle on a circle of radius $r$ is

$$
\ell = \alpha r
\qquad
(\alpha\ \text{in radians}).
$$
