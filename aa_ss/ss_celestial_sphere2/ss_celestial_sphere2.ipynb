## Equatorial coordinate system

<img src="images/celestial_sphere_equator.png" style="width:2.87708in;height:2.22531in" />

If you have ever tried to point out a planet, star, or galaxy to a friend, you will appreciate the difficulty of matching up your line of sight to that of your friend. In this situation, the horizontal coordinate system is a simple and handy set of coordinates with which to identify objects in the sky. Rather than trying to identify a planet to a friend with your friend by trying to match up to your line of sight, you can just say “18 degrees north latitude and 23 degrees west longitude”. Using either tools or their hand, they will then be able to find the object you are trying to describe.

Now suppose you want to tell a classmate where in the sky to look for that astronomical object the following evening. That object may still be at or near the same point in the sky, but your friend’s position on Earth may not match yours, and anyway the position of Earth has changed in the meantime. To overcome these significant hurdles, we need another coordinate system that does not change relative to the fixed stars, or at least changes only in a way that is corrected relatively easily. A system of this type is the [<u>equatorial coordinate system</u>](https://en.wikipedia.org/wiki/Equatorial_coordinate_system). In this system, the Earth is surrounded by the celestial sphere with the fundamental plane passing through Earth’s equator. This plane intersects the celestial sphere in the **celestial equator** (the red circle in the figure below). The coordinate of latitude in degrees above the great circle is called the **declination** (DEC). It is denoted by the lower-case Greek letter delta $\delta$ (or just DEC) and it runs from $- 90{^\circ}$ below the equator to $+ 90{^\circ}$ above.

To fix a coordinate of longitude, we need the analog of the prime meridian. To define it, we need to first discuss the apparent motion of the Sun in the sky. During the day, the Sun rises in the East and sets in the West, but where exactly those points are relative to due East/West depends on the time of year. During the summers in the northern hemisphere, the Sun rises and sets north of the cardinal directions and during the winter it does so south of them. Because of this, there are two days a year when the Sun rises due East and sets due West called the **equinoxes**.

<img src="images/annual_solar_motion.png" style="width:3.44444in;height:3.44444in" />

There are also two days at which the Sun rises and sets as far away from the due East/West as possible. During these days the sun reaches either the highest point in the sky (summer) or the lowest point in the sky (winter) for that year. These days are called the **summer solstice** and **winter solstice**, respectively.

In this problem, you should have discovered that the ecliptic is defined to be the apparent path the Sun traces out across the celestial sphere in one year. This path is tilted relative to the celestial equator at an angle of $23.4{^\circ}$, and it intersects it at antipodal points corresponding to the position of the sun during the equinoxes.

Returning now to our equatorial coordinate system, we define the **primary direction** to be *the direction from Earth to the vernal equinox* (labeled ♈ (Aries) in the figure at right). Starting from this direction, we define a coordinate increasing in the couner-clockwise direction when viewed from the north. This coordinate is called **right ascension** (RA) and is denoted by the Greek letter $\alpha$ (or just RA). It is defined to run from 0 hours to 24 hours with angles further subdivided into minutes (and seconds, *etc*. if needed).

In the previous problem, you were asked to relate the horizontal and ecliptic coordinates of the sun *at noon*. However, as the Earth **rotates** on its axis and **revolves** around the sun, you might guess that the relation between the coordinates is quite complicated, and you would be right. (If you are curious, you can find the transformation rules [<u>here</u>](https://en.wikipedia.org/wiki/Astronomical_coordinate_systems#Equatorial_%E2%86%94_horizontal).)

There are online calculators you can use to convert between the two (cf. *e.g.* [<u>here</u>](https://clearskytonight.com/projects/astronomycalculator/coordinate/ecliptic_equatorial.html)), but this conversion is not usually needed for amateur astronomy. Instead, it is more useful to learn to locate the origin of coordinates in the night sky, and then refer to that whenever you are looking for a specific astronomical object.

Certain combinations of stars have been decreed since antiquity to form **constellations**. There are 88 constellations total, of which 12 are intersected by the ecliptic.
