## Astronomical coordinate systems

Throughout this course we will be faced with the fact that we cannot easily determine the distance to an object that we can see in space: We only see in the two dimensions spanning the sky and have no “depth perception”. This two-dimensional space is called the **celestial sphere**. The two dimensions are called **latitude** (up and down) and **longitude** (side-to-side). Points on the celestial sphere are labeled by two angles called **celestial coordinates**. Sizes of astronomical objects on this sphere are measured in these angles.

<img src="images/earth_latitude_longitude.png" style="width:2.55966in;height:2.00179in" alt="undefined" />

On the surface of the earth, the coordinate for latitude (conventionally denoted by the Greek lower-case letter phi $\phi$) is defined relative to the **equator**: The points of latitude $\phi = 0{^\circ}$ are the equator. There is only one point at a latitude of $\phi = 90{^\circ}$ N (a.k.a. $90{^\circ}$ north-latitude) which is the North Pole. To define the coordinate of longitude, we need a second great circle perpendicular to the equator called the **prime meridian**. For historical reasons, the prime meridian is defined to go through the [<u>Royal Observatory in Greenwich Park</u>](https://en.wikipedia.org/wiki/Royal_Observatory,_Greenwich) in London, England with the angle increasing from $0{^\circ}$ to $180{^\circ}$ east or west.

The previous problem makes it clear that we need to subdivide degrees if we are to be able to navigate to within a precision of hundreds of kilometers. The most common subdivisions are **minutes of arc** and **seconds of arc**. One **arcminute** is $\frac{1}{60}$<sup>th</sup> of a degree. It is denoted $1' \approx 0.0167{^\circ}$. Similarly, one **arcsecond** is is $\frac{1}{60}$<sup>th</sup> of an arcminute and, therefore, $\frac{1}{3600}$<sup>th</sup> of a degree: $1'' \approx 0.0167'\  \approx 2.78 \times 10^{- 4}$ degrees.<sup>1</sup>

## Horizontal coordinate system

<img src="images/horizontal_coordinate_system.png" style="width:3.24458in;height:3.11979in" />

Of course, we are not studying astronomy to learn about navigation, but we use the same ideas to navigate in the night sky as we do under it. To define longitude, we need to decide what to use as the **fundamental plane** (instead of the equatorial plane). There are many options, but a good place to start is to take our own vantage point (from wherever we are on Earth) and the horizon that comes with it. Then instead of, say $45{^\circ}$ north of the equator, we might say $7{^\circ}$ north of the horizon. (Where did I get $7{^\circ}$? Did I just make this up?) This coordinate is called the **altitude**.<sup>2</sup> The point directly overhead is called the **zenith**. It has altitude $90{^\circ}$. To define a coordinate for longitude, we need the analog of the prime meridian. We take this to be given by the plane that *passes through the north pole (and our position) and is orthogonal to the horizon*, in other words, we start by facing north and measure the angle horizontally from there. The convention is to do this in eastward direction. This coordinate is called the **azimuth**. This celestial coordinate system is called the [<u>Horizontal coordinate system</u>](https://en.wikipedia.org/wiki/Horizontal_coordinate_system).

### Astranatomy

It may happen that you are out and about in the world and you notice something that piques your curiosity like a tall tower or an **asterism** you had not noticed before. In these situations, it could be useful to know the angular size of a comparable object. For example, if you know that the tip of your index finger when extended as far from your eyes as possible is roughly $1{^\circ}$, you can use this to estimate that the size of the moon is roughly $0.5{^\circ}$.<sup>3</sup>

>

This can be done alone, but it is easiest to work with a partner. Stand with fully outstretched arm (right or left) so that your middle finger is as far from your face as possible. Close one eye (remember which one it is!) and look down your arm at your finger. (Note that if your finger is as far away from your face as possible, you should be facing sideways at this point.)
>
> a.  Have a partner help you measure the distance from your open eye to the tip of your middle finger.
>
> b.  Measure the width of the nail on your middle finger. Use this and the answer from a. to find the angular size in degrees of your nail when it is this far from your face. Your answer should be around $1{^\circ}$.
>
> c.  Now stretch out your hand and measure the largest distance from the tip of your thumb to the tip of your pinky finger. Have your partner re-measure the distance of your hand to your eye. (It should be slightly less than what you found in part a.) Make note of these two measurements and use them to find the angular size (in degrees) of your outstretched hand.
>
> d.  Repeat the steps for a fist. First measure the angular size of the full width of your fist and then measure knuckle-to-knuckle.

A useful mnemonic is that your finger is around $1{^\circ}$, your fist is around $10{^\circ}$ (so half of it is around $5{^\circ}$), and your outstretched hand is between $15{^\circ}$ and $20{^\circ}$.
